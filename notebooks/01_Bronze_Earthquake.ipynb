### Define date parameters

```python
start_date = "2026-09-11"
end_date = "2026-09-17"
```

### Import Python libraries

```python
import requests
import json
from pathlib import Path
```

### Retrieve the earthquake events

```python
url = (
"https://earthquake.usgs.gov/"
"fdsnws/event/1/query"
)
params = {
"format": "geojson",
"starttime": start_date,
"endtime": end_date
}
response = requests.get(url, params=params, timeout=60)
response.raise_for_status()
data = response.json()
earthquakes = data["features"]
print("Events retrieved:", len(earthquakes))
```

### Write the raw JSON file

```python
folder = Path("/lakehouse/default/Files")
folder.mkdir(parents=True, exist_ok=True)
file_path = folder / f"{start_date}_earthquake_data.json"
with open(file_path, "w") as file:
    json.dump(earthquakes, file)
print("Saved to:", file_path)
```
